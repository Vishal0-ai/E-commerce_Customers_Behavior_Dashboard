```python
labels=['Young Adult','Adult','Middle Age', 'Senior Citizen']
data['Age_group']=pd.qcut(data['Age'], q=4, labels=labels)
```

```python
data[['Age','Age_group']].head()
```

```python
Frequency={ 'Fortnightly': 14,
            'Weekly': 7,
            'Monthly':30,
            'Annually':365,
            'Quarterly':90,
            'Bi-Weekly':14,
            'Every 3 Months':90
          }
data['Purchase_days']=data['Frequency_of_Purchases'].map(Frequency)
```
